Q1 — Basic Dictionary Creation
Create a dictionary called student with keys: name, age, course, city. Print each value.
student = {
"name": "",
"age": ,
"course": "",
"city": ""
}


```python
# Creating dictionary
student = {
    "name": "Ali",
    "age": 21,
    "course": "Software Engineering",
    "city": "Karachi"
}

# Printing each value
print("Name:", student["name"])
print("Age:", student["age"])
print("Course:", student["course"])
print("City:", student["city"])
```

Q3 — Loop Through Dictionary
person = {
"name": "Ali",
"age": 25,
"city": "Karachi"
}
Print keys and values using loop.

```python
person = {
    "name": "Ali",
    "age": 25,
    "city": "Karachi"
}

# Loop through dictionary
for key, value in person.items():
    print(key, ":", value)
```

Q4 — Count Frequency of Numbers
numbers = [1,2,2,3,3,3,4,4,4,4]
Create frequency dictionary.

```python
numbers = [1,2,2,3,3,3,4,4,4,4]

# Create empty dictionary
frequency = {}

# Loop through list
for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

# Print result
print(frequency)
```

Q5 — Word Counter
Take sentence input and count word frequency.

```python
# Taking input from user
sentence = input("Enter a sentence: ")

# Convert sentence into words
words = sentence.split()

# Create empty dictionary
frequency = {}

# Count frequency
for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

# Print result
print(frequency)
```

Q6 — Student Marks & Average
marks = {"Math":80, "English":70, "Physics":90}
Calculate total and average.


```python
marks = {"Math":80, "English":70, "Physics":90}

# Calculate total
total = sum(marks.values())

# Calculate average
average = total / len(marks)

# Print results
print("Total Marks:", total)
print("Average Marks:", average)
```

Q7 — Highest Scoring Student
students = {
"Ali":85,
"Sara":92,
"Ahmed":78,
"Zara":95
}
Find highest scoring student.

```python
students = {
    "Ali":85,
    "Sara":92,
    "Ahmed":78,
    "Zara":95
}

# Find highest scoring student
top_student = max(students, key=students.get)

print("Highest Scoring Student:", top_student)
print("Marks:", students[top_student])
```
